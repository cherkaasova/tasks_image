```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('sar_1.jpg', cv2.IMREAD_GRAYSCALE)
h, w = image.shape
s = min(h, w)
image = cv2.resize(image[(h-s)//2:(h+s)//2, (w-s)//2:(w+s)//2], (512, 512))
plt.imshow(image, cmap='gray')
```

```python
def haar(image, levels):
    c = image.astype(np.float32)
    for l in range(levels):
        n = c.shape[0] >> l
        b = c[:n, :n]
        b = np.hstack(((b[:, 0::2] + b[:, 1::2]) / 2, (b[:, 0::2] - b[:, 1::2]) / 2))
        c[:n, :n] = np.vstack(((b[0::2] + b[1::2]) / 2, (b[0::2] - b[1::2]) / 2))
    return c

def ihaar(coef, levels):
    c = coef.copy()
    for l in range(levels - 1, -1, -1):
        n = c.shape[0] >> l
        k = n // 2
        b = c[:n, :n]
        t = np.empty_like(b)
        t[0::2], t[1::2] = b[:k] + b[k:], b[:k] - b[k:]
        c[:n, :n][:, 0::2], c[:n, :n][:, 1::2] = t[:, :k] + t[:, k:], t[:, :k] - t[:, k:]
    return c

plt.figure(figsize=(18, 6))
for l in range(1, 4):
    c = haar(image, l)
    vis = np.clip(np.abs(c) * 5, 0, 255)
    n = 512 >> l
    vis[:n, :n] = c[:n, :n]
    plt.subplot(1, 3, l), plt.imshow(vis, cmap='gray'), plt.title(str(l) + ' ступень, LL ' + str(n) + 'x' + str(n)), plt.axis('off')
plt.show()
```

```python
coef = haar(image, 3)
print('ошибка восстановления:', np.abs(ihaar(coef, 3) - image).max())
c = coef.copy()
c[np.abs(c) < 5] = 0
c[:64, :64] = coef[:64, :64]
res = np.clip(ihaar(c, 3), 0, 255)
print('обнулено:', round(np.mean(c == 0) * 100, 1), '%   MSE:', round(np.mean((res - image) ** 2), 2))

plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap='gray'), plt.title('исходное'), plt.axis('off')
plt.subplot(122), plt.imshow(res, cmap='gray'), plt.title('после сжатия'), plt.axis('off')
plt.show()
```
